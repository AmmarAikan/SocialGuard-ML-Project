# SocialGuard AI: Fake Social Media Profile Detection Using Machine Learning

**Cybersecurity Business Problem:** Fake social media profiles are commonly used for phishing, impersonation, spam, fake engagement, misinformation, and social engineering. This project builds a machine learning system that classifies user profiles as **Real** or **Fake** using profile-level data and aggregated activity behavior.

## Project Scope in Introduction

**In scope:**
- Merge two Kaggle CSV files into one user-level dataset.
- Perform data discovery, cleaning, EDA, preprocessing, feature engineering, model training, evaluation, model persistence, web deployment, and API deployment.
- Train the six required classification algorithms: Random Forest, Logistic Regression, KNN, SVM, AdaBoost, and XGBoost.
- Provide two prediction modes in deployment: select an existing dataset row index or manually enter profile/activity values.

**Out of scope:**
- Live scraping from real social media accounts.
- Storing or processing real private user accounts.
- Multi-class fake-account family classification. The target here is binary: Real vs Fake.

This notebook follows the required project phases and presents the work as a realistic cybersecurity solution, not only as a checklist for the rubric.

## 1. Load Raw Dataset Files
The Kaggle dataset contains two files: one for user profiles and one for user post/activity records. The project aggregates activities by `user_id` and merges them with user profiles.

```python
profiles = pd.read_csv(DATA_DIR / "raw_user_profiles.csv")
activities = pd.read_csv(DATA_DIR / "raw_user_activities.csv")

print("Profiles shape:", profiles.shape)
print("Activities shape:", activities.shape)
display(profiles.head())
display(activities.head())
```

## 2. Dataset Constraint Check
The rubric requires at least **1,000 records** and at least **10 features**. The raw dataset already satisfies these constraints, and the merged user-level dataset also satisfies them.

```python
df = build_user_level_dataset(save_processed=True)
X, y = create_features_and_target(df)

constraint_check = pd.DataFrame({
    "Check": ["User-level records", "Raw merged columns", "Features used after leakage control", "Target classes"],
    "Value": [df.shape[0], df.shape[1], X.shape[1], y.nunique()],
    "Required": [">= 1000", ">= 10", ">= 10", "2 for binary classification"],
    "Status": ["Pass" if df.shape[0] >= 1000 else "Fail",
               "Pass" if df.shape[1] >= 10 else "Fail",
               "Pass" if X.shape[1] >= 10 else "Fail",
               "Pass" if y.nunique() == 2 else "Fail"]
})
display(constraint_check)
print("Target distribution:")
display(y.value_counts().rename(index={0: "Real", 1: "Fake"}))
```

## 3. Data Cleaning Summary
Cleaning includes duplicate removal, timestamp parsing, missing-value handling through imputers, and removal of identifiers/PII such as username, email, and names.

```python
cleaning_summary = pd.DataFrame({
    "Item": [
        "Profile duplicate user_id rows",
        "Activity duplicate activity_id rows",
        "Missing values in profiles",
        "Missing values in activities",
        "Removed PII/identifier columns",
        "Removed leakage-prone synthetic columns"
    ],
    "Value": [
        profiles.duplicated(subset=["user_id"]).sum(),
        activities.duplicated(subset=["activity_id"]).sum(),
        int(profiles.isna().sum().sum()),
        int(activities.isna().sum().sum()),
        "user_id, username, full_name, first_name, last_name, email",
        ", ".join(LEAKAGE_CONTROL_DROP_COLUMNS)
    ]
})
display(cleaning_summary)
```

### EDA 2: Missing Values by Column

```python
missing = df.isna().sum().sort_values(ascending=False).head(15)
plt.figure(figsize=(9,4))
plt.bar(missing.index, missing.values)
plt.xticks(rotation=45, ha="right")
plt.title("Top Missing Values After Merge")
plt.ylabel("Missing Count")
plt.tight_layout()
plt.show()
display(missing.to_frame("missing_count"))
```

### EDA 3: Profile Completeness Distribution

```python
plt.figure(figsize=(7,4))
plt.hist(df["profile_completeness"].dropna(), bins=25)
plt.title("Distribution of Profile Completeness")
plt.xlabel("Profile Completeness")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

### EDA 4: Followers and Following Box Plots

```python
plt.figure(figsize=(7,4))
plt.boxplot([df["followers_count"].dropna(), df["following_count"].dropna()], labels=["Followers", "Following"])
plt.title("Outlier Detection: Followers vs Following")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

### EDA 5: Correlation Heatmap for Numeric Features

```python
numeric_for_corr = X.select_dtypes(include="number")
corr = numeric_for_corr.corr().fillna(0)
selected_cols = corr.abs().mean().sort_values(ascending=False).head(12).index
small_corr = numeric_for_corr[selected_cols].corr().fillna(0)

plt.figure(figsize=(8,6))
plt.imshow(small_corr, aspect="auto")
plt.colorbar(label="Correlation")
plt.xticks(range(len(selected_cols)), selected_cols, rotation=60, ha="right")
plt.yticks(range(len(selected_cols)), selected_cols)
plt.title("Correlation Heatmap: Top Numeric Features")
plt.tight_layout()
plt.show()
```

### EDA 6: Engagement Rate by Class

```python
real_eng = df.loc[df["is_fake"] == False, "engagement_rate"].replace([np.inf, -np.inf], np.nan).dropna()
fake_eng = df.loc[df["is_fake"] == True, "engagement_rate"].replace([np.inf, -np.inf], np.nan).dropna()
plt.figure(figsize=(6,4))
plt.boxplot([real_eng, fake_eng], labels=["Real", "Fake"], showfliers=False)
plt.title("Engagement Rate by Class")
plt.ylabel("Engagement Rate")
plt.tight_layout()
plt.show()
```

### EDA 7: Dominant Platform Distribution by Class

```python
platform_table = pd.crosstab(df["dominant_platform"], df["is_fake"], normalize="index")
platform_table = platform_table.rename(columns={False: "Real", True: "Fake"})
display(platform_table)
platform_table.plot(kind="bar", figsize=(7,4))
plt.title("Dominant Platform Distribution by Class")
plt.ylabel("Class Proportion")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```

### EDA 8: Outlier Detection with IQR

```python
outlier_cols = ["followers_count", "following_count", "posts_count", "likes_sum", "comments_sum", "shares_sum"]
outlier_rows = []
for col in outlier_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outliers = ((df[col] < lower) | (df[col] > upper)).sum()
    outlier_rows.append({"Feature": col, "Q1": q1, "Q3": q3, "IQR": iqr, "Outlier Count": int(outliers)})
outlier_summary = pd.DataFrame(outlier_rows)
display(outlier_summary)
```

### EDA 9: Numeric Feature Relationship with Target

```python
target_corr = numeric_for_corr.apply(lambda col: col.corr(y)).abs().sort_values(ascending=False).head(12)
plt.figure(figsize=(8,4))
plt.bar(target_corr.index, target_corr.values)
plt.xticks(rotation=45, ha="right")
plt.title("Top Numeric Feature Correlations with Target")
plt.ylabel("Absolute Correlation")
plt.tight_layout()
plt.show()
display(target_corr.to_frame("abs_target_correlation"))
```

# Phase II: Preprocessing and Feature Engineering
This phase includes encoding, scaling, SMOTE, domain-specific feature creation, and PCA.

## Preprocessing Steps Implemented
The project implements the same core deployment ideas used in the example project, but adapted to this dataset:

- **Handling Missing Values:** `SimpleImputer(strategy="median")` for numerical features and `SimpleImputer(strategy="most_frequent")` for categorical features.
- **Encoding:** `OneHotEncoder(handle_unknown="ignore")` for categorical columns such as device and platform.
- **Scaling:** `StandardScaler` for all numerical features.
- **Sampling:** `SMOTE` is applied on the training set only to avoid leakage.
- **Feature Engineering:** domain-specific profile/activity features plus PCA dimensionality reduction.
- **Saved Deployment Artifacts:** preprocessor, scaler, PCA, label encoder, categorical values, feature names, default values, and all trained models.

```python
preprocessor, numeric_features, categorical_features = build_preprocessor(X)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)

print("Numeric features:", len(numeric_features))
print("Categorical features:", categorical_features)
print("Train shape:", X_train.shape, "Test shape:", X_test.shape)

# The actual transformation, SMOTE, and PCA are performed in train_models.py
# to guarantee that preprocessing is fitted only on training data.
```
